   1.Question 1 : Count how many cities are there in each country?


  

sql
USE world;

SELECT
    co.Name AS Country,
    COUNT(ci.ID) AS Number_of_Cities
FROM country co
LEFT JOIN city ci
    ON co.Code = ci.CountryCode
GROUP BY co.Code, co.Name
ORDER BY Number_of_Cities DESC;


 Explanation

* country` contains country information.
* city` contains city information with `CountryCode` linking it to `country.Code`.
* COUNTci.ID` counts the cities belonging to each country.
* LEFT JOIN` ensures countries with no cities in the `city` table are also included.
* GROUP BY` aggregates the count for each country.
* ORDER BY Number_of_Cities DESC` shows countries with the most cities first.


Question 2 : Display all continents having more than 30 countries

  -To display all continents that have **more than 30 countries** in the **World** database, use the following query:

sql
USE world;

SELECT
    Continent,
    COUNT(Code) AS Number_of_Countries
FROM country
GROUP BY Continent
HAVING COUNT(Code) > 30;


Explanation

* COUNT(Code)` counts the number of countries in each continent.
* GROUP BY Continent` groups countries by continent.
* HAVING COUNT(Code) > 30` filters the results to show only continents with more than 30 countries.

 Question 5 : Find the total number of official languages spoken in each continent

   -To find the **total number of official languages spoken in each continent** in the **World** database, join the `country` and `countrylanguage` tables and count the official languages.

sql id="q5official"
USE world;

SELECT
    c.Continent,
    COUNT(cl.Language) AS Total_Official_Languages
FROM country c
JOIN countrylanguage cl
    ON c.Code = cl.CountryCode
WHERE cl.IsOfficial = 'T'
GROUP BY c.Continent
ORDER BY Total_Official_Languages DESC;
```

 Explanation

* `country` contains continent information.
* `countrylanguage` contains languages spoken in each country.
* `IsOfficial = 'T'` filters only official languages.
* `COUNT(cl.Language)` counts official language records for each continent.
* `GROUP BY c.Continent` aggregates the results by continent.

> Note: This query counts official language entries across countries. If the same language is official in multiple countries within a continent, it will be counted multiple times.

If you want the **number of distinct official languages** per continent, use:

sql
USE world;

SELECT
    c.Continent,
    COUNT DISTINCT cl.Language AS Distinct_Official_Languages
FROM country c
JOIN countrylanguage cl
    ON c.Code = cl.CountryCode
WHERE cl.IsOfficial = 'T'
GROUP BY c.Continent
ORDER BY Distinct_Official_Languages DESC;


Question 6 : Find the maximum and minimum GNP for each continent


  -To find the **maximum and minimum GNP for each continent** in the **World** database, use the following query:

```sql id="q6gnp"
USE world;

SELECT
    Continent,
    MAX(GNP) AS Maximum_GNP,
    MIN(GNP) AS Minimum_GNP
FROM country
GROUP BY Continent
ORDER BY Continent;


### Explanation

* `MAX(GNP)` returns the highest GNP among countries in each continent.
* `MIN(GNP)` returns the lowest GNP among countries in each continent.
* `GROUP BY Continent` groups countries by continent.
* `ORDER BY Continent` displays the results alphabetically by continent.

This query will show each continent along with its highest and lowest country GNP values.

To list the continents where the average city population is greater than 200,000, join the country and city tables, calculate the average city population for each continent, and filter the results.

USE world;

SELECT
    co.Continent,
    AVG(ci.Population) AS Average_City_Population
FROM country co
JOIN city ci
    ON co.Code = ci.CountryCode
GROUP BY co.Continent
HAVING AVG(ci.Population) > 200000
ORDER BY Average_City_Population DESC;
Explanation
JOIN links cities to their respective countries.
AVG(ci.Population) calculates the average population of all cities within each continent.
GROUP BY co.Continent groups the data by continent.
HAVING AVG(ci.Population) > 200000 filters continents whose average city population exceeds 200,000.
ORDER BY Average_City_Population DESC sorts the results from highest to lowest average city population.
Question 9 : Find the total population and average life expectancy for each continent, ordered by average life
expectancy descending

To find the total population and average life expectancy for each continent, ordered by average life expectancy in descending order, use:

USE world;

SELECT
    Continent,
    SUM(Population) AS Total_Population,
    AVG(LifeExpectancy) AS Average_Life_Expectancy
FROM country
GROUP BY Continent
ORDER BY Average_Life_Expectancy DESC;
Explanation
SUM(Population) calculates the total population of all countries in each continent.
AVG(LifeExpectancy) computes the average life expectancy for countries within each continent.
GROUP BY Continent groups the records by continent.
ORDER BY Average_Life_Expectancy DESC sorts continents from highest to lowest average life expectancy.

Note: AVG() automatically ignores NULL values in the LifeExpectancy column.

Question 10 : Find the top 3 continents with the highest average life expectancy, but only include those where
the total population is over 200 million

To find the top 3 continents with the highest average life expectancy, considering only continents whose total population exceeds 200 million, use:

USE world;

SELECT
    Continent,
    SUM(Population) AS Total_Population,
    AVG(LifeExpectancy) AS Average_Life_Expectancy
FROM country
GROUP BY Continent
HAVING SUM(Population) > 200000000
ORDER BY Average_Life_Expectancy DESC
LIMIT 3;
Explanation
SUM(Population) calculates the total population for each continent.
AVG(LifeExpectancy) computes the average life expectancy of countries within each continent.
GROUP BY Continent groups countries by continent.
HAVING SUM(Population) > 200000000 filters continents with a total population greater than 200 million.
ORDER BY Average_Life_Expectancy DESC ranks continents by average life expectancy.
LIMIT 3 returns only the top 3 continents.
